```python
class LRIfromscratch(nn.Module):
  def __init__(self, lr, num, sigma=0.01):
    super.__init__()
    self.lr = lr
    self.num = num
    self.sigma = sigma
    self.w=nn.Parameter(torch.normal(0, sigma, (num, 1)))
    self.b=nn.Parameter(torch.zeros(1))

  def forward(self, x):
    return torch.matmul(x, self.w) + self.b
  def loss(self, y_hat, y):
    return (0.5*(y_hat - y)**2).mean()

```
